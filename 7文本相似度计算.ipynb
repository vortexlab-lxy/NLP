词袋模型（Bag of Words）

```python
from sklearn.feature_extraction.text import CountVectorizer

corpus = ["我喜欢自然语言处理", "我爱学习NLP技术", "文本相似度计算很有趣"]

vectorizer = CountVectorizer(analyzer="char", ngram_range=(1, 2), lowercase=False)
bow_matrix = vectorizer.fit_transform(corpus)

print("词频矩阵:", bow_matrix.toarray())
print("特征名称:", vectorizer.get_feature_names_out())
```

TF-IDF 方法

```python
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = ["我喜欢自然语言处理", "我爱学习NLP技术", "文本相似度计算很有趣"]

corpus_tfidf = TfidfVectorizer(analyzer="char", ngram_range=(1, 2), lowercase=False)
corpus_tfidf_matrix = corpus_tfidf.fit_transform(corpus)

print("TF-IDF矩阵:", corpus_tfidf_matrix.toarray())
print("特征名称:", corpus_tfidf.get_feature_names_out())
```

Word2Vec 词向量

```python
from gensim.models import Word2Vec

sentences = [
    ['我', '喜欢', '自然语言处理'],
    ['我', '爱', '学习', 'NLP', '技术'],
    ['文本', '相似度', '计算', '很', '有趣']
]

model = Word2Vec(sentences, vector_size=100, window=5, min_count=1, workers=4)

# 获取词向量
vector = model.wv['自然语言处理']
print("'自然语言处理' 的词向量（前10维）:", vector[:10])
```

句子向量计算

```python
import numpy as np

def sentence_vector(sentence, model):
    vectors = [model.wv[word] for word in sentence if word in model.wv]
    return np.mean(vectors, axis=0) if vectors else np.zeros(model.vector_size)

# 计算两个句子的向量
sentence_vec1 = sentence_vector(['我', '喜欢', '自然语言处理'], model)
sentence_vec2 = sentence_vector(['我', '爱', 'NLP'], model)

print("句子1向量（前10维）:", sentence_vec1[:10])
print("句子2向量（前10维）:", sentence_vec2[:10])
```

BERT 相似度计算

```python
from pathlib import Path

import torch
from transformers import AutoModel, AutoTokenizer
# ============================================================
# 1. 指定本地中文BERT模型目录
#
# 注意：
# 从你的截图看，文件夹名称是 bertbert-base-chinese
# ============================================================
model_dir = Path(r"D:\11\NLP\data\bertbert-base-chinese")
# ============================================================
# 2. 检查模型文件夹
# ============================================================
if not model_dir.is_dir():
    raise FileNotFoundError(f"模型文件夹不存在：{model_dir}\n""请检查左侧项目目录中的真实文件夹名称。")

print("本地模型目录：")
print(model_dir)
print("\n文件夹中的文件：")
for file_path in model_dir.iterdir():
    print(file_path.name)
# ============================================================
# 3. 检查必要文件
# ============================================================
required_basic_files = [ "config.json","vocab.txt"]
missing_files = []
for file_name in required_basic_files:
    if not (model_dir / file_name).is_file():
        missing_files.append(file_name)

# 模型权重可能是两种格式之一
has_model_weights = ((model_dir / "model.safetensors").is_file() or(model_dir / "pytorch_model.bin").is_file())

if missing_files:
    raise FileNotFoundError("中文BERT文件不完整，缺少：\n" + "\n".join(missing_files))

if not has_model_weights:
    raise FileNotFoundError(
        "模型文件夹中没有找到模型权重。\n"
        "至少需要下面其中一个文件：\n"
        "model.safetensors\n"
        "pytorch_model.bin\n\n"
        f"当前目录：{model_dir}"
    )

print("\n模型基础文件检查成功！")

# ============================================================
# 4. 完全从本地加载分词器
# ============================================================
tokenizer = AutoTokenizer.from_pretrained(str(model_dir), local_files_only=True,use_fast=True)
# ============================================================
# 5. 完全从本地加载BERT模型
# ============================================================
model = AutoModel.from_pretrained(str(model_dir),local_files_only=True)
# 切换为推理模式
model.eval()
print("中文BERT分词器加载成功！")
print("中文BERT模型加载成功！")

# ============================================================
# 6. 输入中文文本
# ============================================================
text = "这是一个示例句子。"

# ============================================================
# 7. 文本编码
# ============================================================
inputs = tokenizer(text,return_tensors="pt")

# ============================================================
# 8. 查看分词结果
# ============================================================
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

print("\n原始文本：")
print(text)
print("\n分词结果：")
print(tokens)
print("\ninput_ids：")
print(inputs["input_ids"])
print("\nattention_mask：")
print(inputs["attention_mask"])

# ============================================================
# 9. BERT前向传播
# ============================================================
with torch.no_grad():
    outputs = model(**inputs)

# ============================================================
# 10. 获取最后一层隐藏状态
#
# 注意：正确名称是 last_hidden_state
# 不是 last_hidden_states
# ============================================================
last_hidden_states = outputs.last_hidden_state

# ============================================================
# 11. 输出结果
# ============================================================
print("\n输入Token数量：")
print(inputs["input_ids"].shape[1])
print("\n最后一层隐藏状态形状：")
print(last_hidden_states.shape)

# ============================================================
# 12. 查看每个Token对应的向量
# ============================================================

print("\n每个Token对应的向量形状：")
for index, token in enumerate(tokens):
    token_vector = last_hidden_states[0, index]
    print(
        f"{index:<3}"
        f"{token:<12}"
        f"{token_vector.shape}"
    )

# ============================================================
# 13. 查看[CLS]向量
# ============================================================
cls_vector = last_hidden_states[0, 0]
print("\n[CLS]向量形状：")
print(cls_vector.shape)
print("\n[CLS]向量前10维：")
print(cls_vector[:10])
```

补充 1：Jaccard 相似度

```python
def jaccard_similarity(text1, text2):
    # 分词（简单按字分割）
    set1 = set(text1)
    set2 = set(text2)
    intersection = len(set1 & set2)
    union = len(set1 | set2)
    return intersection / union if union > 0 else 0

# 测试
text1 = "我喜欢自然语言处理"
text2 = "我爱学习NLP技术"
print(f"Jaccard相似度: {jaccard_similarity(text1, text2):.4f}")
```

补充 3：使用 Sentence-BERT 计算语义相似度

```python
from sentence_transformers import SentenceTransformer, models, util

# 你的本地模型路径
local_model_path = r"D:\11\NLP\data\paraphrase-multilingual-MiniLM-L12-v2"

# 方法：手动组合模型模块（避开自动解析的 Bug）
# 1. 加载 Transformer 底座（它会自动去读你文件夹里的 config.json 和 model.safetensors）
word_embedding_model = models.Transformer(local_model_path)

# 2. 从这里直接获取正确的维度（这就解决了找不到 word_embedding_dimension 的问题）
dim = word_embedding_model.get_word_embedding_dimension()

# 3. 添加池化层（把词向量变成句子向量）
pooling_model = models.Pooling(dim)

# 4. 组合为最终的 SentenceTransformer
model = SentenceTransformer(modules=[word_embedding_model, pooling_model])

# -------------------------------------------------------------
# 下面是你原本的代码，保持不变即可：

# 待比较的句子
sentences = [
    "苹果发布新款iPhone手机",
    "苹果公司推出最新智能手机",
    "微软公布季度财报"
]

# 生成句子向量
embeddings = model.encode(sentences)

# 计算相似度
cos_sim = util.cos_sim(embeddings, embeddings)

print("相似度矩阵:\n", cos_sim)
```

补充 4：完整的文本相似度对比函数

```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def compare_texts(text1, text2):
    results = {}

    tfidf_vectorizer = TfidfVectorizer(analyzer="char", ngram_range=(1, 2), lowercase=False)
    tfidf_vectors = tfidf_vectorizer.fit_transform([text1, text2])
    results["TF-IDF余弦相似度"] = cosine_similarity(tfidf_vectors[0:1], tfidf_vectors[1:2])[0][0]

    set1, set2 = set(text1), set(text2)
    results["Jaccard相似度"] = len(set1 & set2) / len(set1 | set2) if len(set1 | set2) > 0 else 0

    count_vectorizer = CountVectorizer(analyzer="char", ngram_range=(1, 2), lowercase=False)
    count_vectors = count_vectorizer.fit_transform([text1, text2])
    results["词袋余弦相似度"] = cosine_similarity(count_vectors[0:1], count_vectors[1:2])[0][0]

    return results

text_a = "苹果发布新款iPhone手机"
text_b = "苹果公司推出最新智能手机"
results = compare_texts(text_a, text_b)

for method, score in results.items():
    print(f"{method}: {score:.4f}")
```
